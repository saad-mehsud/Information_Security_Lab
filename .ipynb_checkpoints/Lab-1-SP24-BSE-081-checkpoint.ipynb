#### **Lab Task 1:**
Write a program that prompts the user to input an integer and then outputs the number with
the digitsreversed. For example, if the input is 12345, the output should be 54321.

```python
num = int(input("Enter an integer: "))

# Handle negative numbers
sign = -1 if num < 0 else 1
num = abs(num)

reversed_num = 0
while num > 0:
    digit = num % 10
    reversed_num = reversed_num * 10 + digit
    num //= 10

print("Reversed number:", sign * reversed_num)
```

#### **Lab Task 2**:
Write a program that reads a set of integers, and then prints the sum of the even and odd integers.

```python
n = int(input("Enter how many integers you want to input: "))

even_sum = 0
odd_sum = 0

for _ in range(n):
    num = int(input("Enter an integer: "))
    if num % 2 == 0:
        even_sum += num
    else:
        odd_sum += num

print("Sum of even integers:", even_sum)
print("Sum of odd integers:", odd_sum)
```

#### **Lab Task 3**:
Fibonacci series is that when you add the previous two numbers the next number is formed.
You haveto start from 0 and 1.
E.g. 0+1=1 → 1+1=2 → 1+2=3 → 2+3=5 → 3+5=8 → 5+8=13

```python
n = int(input("Enter the number of terms: "))

a, b = 0, 1
print("Fibonacci series:")

for _ in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

#### **Lab Task 4**:
Write a Python code to accept marks of a student from 1-100 and display the grade
according to the following formula.
- Grade F if marks are less than 50
- Grade E if marks are between 50 to 60
- Grade D if marks are between 61 to 70
- Grade C if marks are between 71 to 80
- Grade B if marks are between 81 to 90
- Grade A if marks are between 91 to 100

```python
marks = int(input("Enter marks (1-100): "))

if marks < 50:
    grade = "F"
elif marks <= 60:
    grade = "E"
elif marks <= 70:
    grade = "D"
elif marks <= 80:
    grade = "C"
elif marks <= 90:
    grade = "B"
elif marks <= 100:
    grade = "A"
else:
    grade = "Invalid marks"

print("Grade:", grade)
```

#### **Lab Task 5**:
Write a program that takes a number from user and calculate the factorial of that number.

```python
num = int(input("Enter a number: "))

if num < 0:
    print("Factorial does not exist for negative numbers")
elif num == 0:
    print("Factorial of 0 is 1")
else:
    factorial = 1
    for i in range(1, num + 1):
        factorial *= i
    print(f"Factorial of {num} is {factorial}")
```
