## Lab Activity 1: Substitution Ciphers
### The substitution cipher simply substitutes one letter in the alphabet for another based upon a cryptovariable. The substitution involves shifting positions in the alphabet. This includes the Caesar cipher and ROT-13, which will be covered shortly. Examine the following example: 
### Plaintext: WE HOLD THESE TRUTHS TO BE SELF-EVIDENT, THAT ALL MEN ARE CREATED EQUAL.
### Ciphertext: ZH KROG WKHVH WUXWKV WR EH VHOI-HYLGHQW, WKDW DOO PHQ DUH FUHDWHG HTXDO. The Python syntax to both encrypt and decrypt a substitution cipher is presented next.This example shows the use of

```python
key = "abcdefghijklmnopqrstuwvxyz"

def encode(n,plaintext):
    result = ""
    for l in plaintext.lower():
        try:
            i = (key.index(l) + n) %26
            result+=key[i]
        except ValueError :
            result += l
    return result

def decode(n,encodeText):
    result = ""
    for l in encodeText.lower():
        try:
            i = (key.index(l) - n) %26
            result+=key[i]
        except ValueError :
            result += l
    return result    
keyText = "WE HOLD THESE TRUTHS TO BE SELF-EVIDENT, THAT ALL MEN ARE CREATED EQUAL."
cipherText = encode(13,keyText)
print(cipherText)
decodedText = decode(13,cipherText)
print(decodedText)
```
