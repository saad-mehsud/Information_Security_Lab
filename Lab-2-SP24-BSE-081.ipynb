### **Activity 2:**
##### A palindrome is a string which is same read forward or backwards. <br> For example: "dad" is the same in forward or reverse direction.<br> Another example is "aibohphobia" which literally means, an irritable fear of palindromes.<br> Write a function in python that receives a string and returns True if that string is a palindrome and False otherwise.<br> Remember that difference between upper and lower case characters are ignored during this determination.
----

```python
def is_Palindrome(word):
    temp = word [::-1]
    if(temp.capitalize() == word.capitalize()):
        return True
    else:
        return False
print(is_Palindrome("zaza"))
```

### **Activity 3:**
##### Imagine two matrices given in the form of 2D lists as under; a = [[1, 0, 0], [0, 1, 0], [0, 0, 1] ] <br>b = [[1, 2, 3], [4, 5, 6], [7, 8, 9] ] Write a python code that finds another matrix/2D list that is a product of and b, i.e., C=a*b
---

```python
def matrixMultiplication(mat1,mat2):
    solutionMatrix =[]
    for indRow in range (3):
        solutionMatrix.append([])
        for indColumn in range (3):
            solutionMatrix[indRow].append(0)
            for indAux in range (3):
                solutionMatrix[indRow][indColumn] += mat1[indRow][indAux] * mat2[indAux][indColumn]
                

    return solutionMatrix
a = [[1, 0, 0], [0, 1, 0], [0, 0, 1] ]
b = [[1, 2, 3], [4, 5, 6], [7, 8, 9] ] 
c = matrixMultiplication(a,b)
print(c)
    
```
